## cap_esacontactonboarding_clean
Filters out ghost rows (missing primary key) from the Dataverse source 
and writes clean data to CRM_Monitoring lakehouse.

```python
from pyspark.sql.functions import col

# Clear cache
spark.catalog.clearCache()

# Read fresh
df_full = spark.table("dataverse_esacontact_cds2_workspace_a94a4bf848e144bba608bb2eb51cbe.cap_esacontactonboarding")

# Filter using isNotNull on the key field
df_clean = df_full.filter(col("cap_esacontactonboardingid").isNotNull())

print(f"Clean rows: {df_clean.count()}")

df_clean.write.format("delta").mode("overwrite").saveAsTable("crm_monitoring.cap_esacontactonboarding_clean")

print("Done")
```
